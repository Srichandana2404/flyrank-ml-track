## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Fixed rules like “refresh every 6 months” or “update pages with CTR below 5%” are too simplistic. Engagement patterns vary by page type, audience, and seasonality. For example, some pages drop quickly after a campaign ends, while evergreen pages hold steady for years. An if‑statement can’t capture these messy, overlapping signals. ML can learn from multiple features — impressions, CTR trends, last refresh date, and content category — to predict which pages are truly decaying. This makes the refresh decision adaptive instead of rigid.
